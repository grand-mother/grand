    Calculate geomagnetic field at a given location and obstime. 'WMMHR' geomagnetic model is used as default. 

    Using the 'geomagnetic' module, the magnetic field can be defined in two different ways (see below) depending on the input location. The default geomagnetic model ('WMMHR') and default obstime ('2026-12-31') are used if not given.

        BfieldRepresentation(x: Union[Number, np.ndarray] = None,           # X-component of magnetic field
                             y: Union[Number, np.ndarray] = None,           # Y-component of magnetic field
                             z: Union[Number, np.ndarray] = None,           # Z-component of magnetic field
                             declination: Union[Number, np.ndarray] = None, # declination angle of magnetic field
                             inclination: Union[Number, np.ndarray] = None, # inclination angle of magnetic field
                             location: Geodetic = None,  # location where magnetic field has been computed
                             orientation: str   = 'NED', # orientation of the coordinate frame. 'NED' is the default orientation of WMMHR.
                             magnetic: bool     = False, # shift orientation by magnetic declination?
                             parent = None,  # store Geomagnet instance to retrieve model and obstime info required for 'to_grandcs' method.
                            )
        Geomagnet(model: str= 'IGRF', # IGRF, WMM, WMMHR
                  obstime: Union[str, datetime.date] = '2026-12-31',
                  latest_epoch: int     = 2025)

        1. Define an instance of geomagnetism with a model and observation time.
            geo = Geomagnet(model='WMMHR', obstime='2026-12-31') # default values are used if input is not given
            
        2. Location can be given as ECEF, Geodetic, LTP, or GRAND coordinate system. 'x' below can be in ECEF, Geodetic, LTP, or GRANDCS.
            loc = Geodetic(latitude=[..], longitude=[..], height=[..], reference="Geoid")
            B = geo.field(location=loc) or simply geo.field(loc)
            
        3. Location can be directly ßgiven as latitude, longitude, and height.
            B = geo.field(latitude=27, longitude=85, height=1300, reference="Ellipsoid") # Number, list, or np.ndarray is accepted
            
    Error will occur if location is not provided. For observation time (obstime), provide time in isoformat ('2026-12-31') or in datetime.date(2026, 12, 31). If obstime is not provided, a default date ('2026-12-31') is used. After the magnetic field is computed, you can get the required quantity by:
        B.x
        B.y
        B.z
        B.declination
        B.inclination
        B.location
        B.to_ltp("NWU") # from "NED" to "NWU" orientation at each location
        B.to_grandcs()  # from given orientation to the GRANDCS.

```python
from grand.geo.geomagnet import Geomagnet, Geodetic

import numpy as np
import datetime
```

```python
# Define a geomagnetic model at a given observation time. This object can be used to compute geomagnetic field at any locations.
geo = Geomagnet(model='IGRF', obstime='2026-12-31') #model= IGRF, WMM, WMMHR

lat = 27   # Number, list, or np.ndarray is accepted
lon = 85   # Number, list, or np.ndarray is accepted
hei = 1300 # Number, list, or np.ndarray is accepted
loc = Geodetic(latitude=lat, longitude=lon, height=hei, reference="Ellipsoid")
```

```python
# Input location can be given as latitude, longitude, and height. Number, list, or np.ndarray is accepted
B1  = geo.field(latitude=27, longitude=85, height=1300, reference="Ellipsoid") 

# Provide location either in ECEF, Geodetic, LTP, or GRANDCS
B2  = geo.field(loc)


print("")
print("-- Magneitc Field (B)")
print(B1.x, B1.y, B1.z, B1.norm(), type(B1))
print(B2.x, B2.y, B2.z, B1.norm(), type(B2))
print("-- Location")
print(B1.location, type(B1.location))
print("--")
print(
    "Declination [deg]:", B1.declination,
    ",  Inclination [deg]:", B1.inclination
)
print("-- Orientation --")
print(B1.orientation)
print("-- Change coordinate frame --")
B1nwu = B1.to_ltp("NWU") # from "NED" to "NWU" orientation at each location
B1gcs = B1.to_grandcs()  # from given orientation to the GRANDCS.
print(B1nwu.x, B1nwu.y, B1nwu.z, B1nwu.norm(), B1nwu.orientation)
print(B1gcs.x, B1gcs.y, B1gcs.z, B1gcs.norm(), B1gcs.orientation)

print("-- Observation time")
print(B1.parent.obstime, type(B1.parent.obstime))
print("-- Model")
print(B1.parent.model, type(B1.parent.model))
print("....................\n")
```
